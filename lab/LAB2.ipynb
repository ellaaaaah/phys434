PROBLEM 1

QUESTION A

Average Cosmic Ray Background in 1 Day (X): 4

Number of Gamma Rays Emitted in 1 Day (Y): 6

```python
x = 4

limits = np.arange(0, 20)
prob = stats.poisson.pmf(limits, x)

fig, ax = plt.subplots(1, 1)
ax.step(limits, prob, where = 'mid')
plt.title('Num of Rays vs Probability')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Probability')
plt.show()
```

```python
day2 = np.convolve(prob, prob, mode = 'full')
day3 = np.convolve(day2, prob, mode = 'full')

fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day2), 1), day2)
ax.step(np.arange(0, len(day3), 1), day3)
plt.title('Num of Rays vs Probability')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Probability')
plt.show()
```

The probability distribution becomes wider and shorter as more days are added.

QUESTION B

```python
day2 = np.convolve(prob, prob, mode = 'full')
day3 = np.convolve(day2, prob, mode = 'full')
day4 = np.convolve(day3, prob, mode = 'full')
day5 = np.convolve(day4, prob, mode = 'full')

fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day2), 1), day2)
ax.step(np.arange(0, len(day3), 1), day3)
ax.step(np.arange(0, len(day4), 1), day4)
ax.step(np.arange(0, len(day5), 1), day5)
plt.title('Num of Rays vs Probability')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Probability')
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day5), 1), day5)
ax.set_yscale('log')
plt.title('Num of Rays vs Probability')
plt.xlabel('Number of Gamma Rays Emitted')
plt.ylabel('LOG(Probability)')
plt.show()
```

After a number of convolutions, the distribution appears to be Gaussian, however, it remains a Poisson distribution, as observable when plotting semilog-y, because it is skewed to the left, and not a perfect upside-down parabola. This makes sense conceptually because summing two Poisson distributions does not change that the underlying data is still organized in a Poisson distribution, thus producing another Poisson distribution.

QUESTION C

As each day passes, the number of bins should increase appropriately (by dividing by the day number), and to keep the height from inaccurately shrinking as well, the convolution is multiplied by the day number. This is because the y axis, in this case, is probability density, not just probability.

```python
fig, ax = plt.subplots(1, 1)
ax.step(limits, prob, where = 'mid')
ax.step(np.arange(0, len(day2), 1) / 2, day2 * 2)
ax.step(np.arange(0, len(day3), 1) / 3, day3 * 3)
ax.step(np.arange(0, len(day4), 1) / 4, day4 * 4)
ax.step(np.arange(0, len(day5), 1) / 5, day5 * 5)
plt.title('Num of Rays vs Probability Density')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Probability Density')
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day2), 1) / 2, day2 * 2)
ax.step(np.arange(0, len(day3), 1) / 3, day3 * 3)
ax.step(np.arange(0, len(day4), 1) / 4, day4 * 4)
ax.step(np.arange(0, len(day5), 1) / 5, day5 * 5)
ax.set_yscale('log')
plt.title('Num of Rays vs Probability Density')
plt.xlabel('Number of Gamma Rays Emitted')
plt.ylabel('LOG(Probability Density)')
plt.show()
```

As the number of days increases, the distribution becomes taller and thinner. This, interestingly, is the opposite of what was observed in part B, where the distribution became shorter and wider (however, it is important to note the change in y axis labels). However, this is still backed up by the Central Limit Theorem because when averaged enough times the probability distribution becomes approximately normal. This can be tested by using a semilog-y plot, as above. As you can see, the semilog-y plot is slowly approaching an upside down parabola (though not observable within 5 days), as the days averaged are increased.

```python
x = 4

limits = np.arange(0, 20)
prob = stats.rayleigh.pdf(limits, x)

day2 = np.convolve(prob, prob, mode = 'full')
day3 = np.convolve(day2, prob, mode = 'full')
day4 = np.convolve(day3, prob, mode = 'full')
day5 = np.convolve(day4, prob, mode = 'full')

fig, ax = plt.subplots(1, 1)
ax.step(limits, prob, where = 'mid')
ax.step(np.arange(0, len(day2), 1) / 2, day2 * 2, where = 'mid')
ax.step(np.arange(0, len(day3), 1) / 3, day3 * 3, where = 'mid')
ax.step(np.arange(0, len(day4), 1) / 4, day4 * 4, where = 'mid')
ax.step(np.arange(0, len(day5), 1) / 5, day5 * 5, where = 'mid')
plt.title('Rayleigh Distribution PDF')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Probability Density')
plt.show()
```

The distribution becomes taller and thinner as the number of observing intervals averaged increases.

QUESTION B

```python
fig, ax = plt.subplots(1, 1)
ax.step(limits, prob, where = 'mid')
ax.step(np.arange(0, len(day2), 1) / 2, day2 * 2, where = 'mid')
ax.step(np.arange(0, len(day3), 1) / 3, day3 * 3, where = 'mid')
ax.step(np.arange(0, len(day4), 1) / 4, day4 * 4, where = 'mid')
ax.step(np.arange(0, len(day5), 1) / 5, day5 * 5, where = 'mid')
plt.title('Rayleigh Distribution PDF')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('LOG(Probability Density)')
ax.set_yscale('log')
plt.show()
```

The distribution becomes less wide and taller as you average over more observing intervals, for example, five, as above. To evaluate if it approaches a Gaussian distribution you use a semilog-y plot. The first 5 days did not appear as an upside-down parabola, so I expand the intervals to 320:

```python
fig, ax = plt.subplots(1, 1)

day6 = np.convolve(day5, prob, mode = 'full')
day7 = np.convolve(day6, prob, mode = 'full')
day8 = np.convolve(day7, prob, mode = 'full')
day9 = np.convolve(day8, prob, mode = 'full')
day10 = np.convolve(day9, prob, mode = 'full')
day20 = np.convolve(day10, day10, mode = 'full')
day40 = np.convolve(day20, day20, mode = 'full')
day80 = np.convolve(day40, day40, mode = 'full')
day160 = np.convolve(day80, day80, mode = 'full')
day320 = np.convolve(day160, day160, mode = 'full')

ax.step(limits, prob, where = 'mid')
ax.step(np.arange(0, len(day10), 1) / 10, day10 * 10, where = 'mid')
ax.step(np.arange(0, len(day20), 1) / 20, day20 * 20, where = 'mid')
ax.step(np.arange(0, len(day40), 1) / 40, day40 * 40, where = 'mid')
ax.step(np.arange(0, len(day80), 1) / 80, day80 * 80, where = 'mid')
ax.step(np.arange(0, len(day160), 1) / 160, day160 * 160, where = 'mid')
ax.step(np.arange(0, len(day320), 1) / 320, day320 * 320, where = 'mid')
plt.title('Rayleigh Distribution PDF')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('LOG(Probability Density)')
ax.set_yscale('log')
plt.show()
```

```python
day640 = np.convolve(day320, day320, mode = 'full')
day1280 = np.convolve(day640, day640, mode = 'full')

fig, ax = plt.subplots(1, 1)
ax.step(limits, prob, where = 'mid')
ax.step(np.arange(0, len(day320), 1) / 320, day320 * 320, where = 'mid')
ax.step(np.arange(0, len(day1280), 1) / 1280, day1280 * 1280, where = 'mid')
plt.title('Rayleigh Distribution PDF')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Probability Density')
ax.set_yscale('log')
plt.show()
```

The semilog-y plot shows that even after 320 intervals, the distribution is still slightly skewed to the left, so in a sense, it is appearing to slowly approach a Gaussian distribution. However, the harsh line around 5 prevents it from fully being Gaussian. Upping the number to 1280 intervals, reveals that the distribution is also slowly shrinking as perhaps not all of the data was encompassed by our original limits of 0 to 20, thus I am going to stop upping the intervals and conclude that any approach towards Gaussian is occuring so slowly as to likely be in the thousands range, or is not attainable.

QUESTION B

Statistical Question: How does the value of the background pixels change when attempting to locate a signal in 10,000 pixels?

As the number of pixels increases, the Law of Large Numbers states the distribution will approach a normal distribution, thus the standard deviation will change from 3 to 1, for a ratio of 1/3.

```python
g0 = stats.norm.rvs(loc = 0, scale = 3, size = 100)

fig, ax = plt.subplots(1, 1)
ax.hist(g0, 20, density = True)
plt.title('PDF of 100 Pixels')
plt.xlabel('Bachground Value')
plt.ylabel('Probability Density')
plt.show()
```

```python
g0 = stats.norm.rvs(loc = 0, scale = 3, size = 10)
g = stats.norm.rvs(loc = 0, scale = 3, size = 10000)

fig, ax = plt.subplots(1, 1)
ax.hist(g, 50, density = True)
plt.title('PDF of 10000 Pixels')
plt.xlabel('Bachground Value')
plt.ylabel('Probability Density')
plt.show()
```

As you can see, as the number of samples increases, the distribution becomes more normal.
