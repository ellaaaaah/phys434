```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats
```

PROBLEM 1

QUESTION A

Average Cosmic Ray Background in 1 Day: 4

Number of Gamma Rays Emitted in 1 Day: 6

```python
x = 4

limits = np.arange(0, 20)
prob = stats.poisson.pmf(limits, x)

fig, ax = plt.subplots(1, 1)
ax.step(limits, prob, where = 'mid')
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Density')
plt.show()
```

```python
day2 = np.convolve(prob, prob, mode = 'full')
day3 = np.convolve(day2, prob, mode = 'full')

fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day2), 1), day2)
ax.step(np.arange(0, len(day3), 1), day3)
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Density')
plt.show()
```

QUESTION B

```python
day2 = np.convolve(prob, prob, mode = 'full')
day3 = np.convolve(day2, prob, mode = 'full')
day4 = np.convolve(day3, prob, mode = 'full')
day5 = np.convolve(day4, prob, mode = 'full')

fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day2), 1), day2)
ax.step(np.arange(0, len(day3), 1), day3)
ax.step(np.arange(0, len(day4), 1), day4)
ax.step(np.arange(0, len(day5), 1), day5)
plt.xlabel('Number of Rays Emitted')
plt.ylabel('Density')
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.step(np.arange(0, len(day5), 1), day5)
ax.set_yscale('log')
plt.xlabel('Number of Gamma Rays Emitted')
plt.ylabel('Density')
plt.show()
```

After a number of convolutions, the distribution appears to be Gaussian, however, it remains a Poisson distribution, as observable when plotting semilog-y, because it is skewed to the left, and not a perfect upside-down parabola.
