QUESTION 2

```python
# generate exponential distribution
exp_dist = stats.expon.rvs(loc = 0, scale = 1, size = 100000)
```

```python
# plot exponential distribution
fig, ax = plt.subplots(1, 1)
ax.hist(exp_dist, 50, density = True)
x = np.linspace(0, 7, 10000)
ax.plot(x, stats.expon.pdf(x, loc = 0, scale = 1))
plt.xlim([0, 7])
plt.xlabel("Number")
plt.ylabel("Density")
plt.show()
```

```python
# semilog (y) plot exponential distribution
fig, ax = plt.subplots(1, 1)
ax.hist(exp_dist, 50, density = True)
plt.yscale('log')
x = np.linspace(0, 13, 10000)
ax.plot(x, stats.expon.pdf(x, loc = 0, scale = 1))
plt.xlim([0, 13])
plt.xlabel("Number")
plt.ylabel("Density")
plt.show()
```

QUESTION 3

My Statistical Question:

Given that my signal-free data follows an exponential distribution, what is the probability of a given data point being equivalent to or below a value of 3?

The Integral Equivalent:
$\int_{0}^{3}e^{x}dx$

```python
# calculating the probability the background produced the signal
stats.expon.cdf(x = 3, loc = 0, scale = 1)
```

```python
# converting probability into an equivalent sigma
stats.expon.ppf(0.950212931632136)
```

```python
# selecting a value of 0.01:
stats.expon.cdf(x = 0.01, loc = 0, scale = 1)
```

```python
# selecting a value of 0.01:
stats.expon.ppf(0.009950166250831945)
```

```python
#selecting a value of 8:
stats.expon.cdf(x = 8, loc = 0, scale = 1)
```

```python
# selecting a value of 8:
stats.expon.ppf(0.9996645373720975)
```

The most notable pattern I observed was that as the value increases, the probability increases. This makes sense because, looking at the plot of the exponential distribution, as the value increases along the x axis, more of the distribution is encompassed by the "less than or equivalent to" characteristic of the statistical question. Though expected due to the nature of the graph, it is still interesting to see how rapidly the distribution is encompassed with each increasing value in the lower numbers (for example the gap in probability between 0.01 - 3 is much larger than that of 3 - 8, despite the smaller numerical gap).
