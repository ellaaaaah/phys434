```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

plt.rcParams["figure.figsize"] = (15, 10)
```

To change the central value of the Gaussian distribution, you change the value of "loc", in this case, to 3.0. To modify the width, change the "scale" value to the desired value.

```python
d = stats.norm.rvs(loc = 3.0, scale = .005, size = 1000000)
```

Next, ensure you modify the x limits and x value (for the line) to center your new central value, as well as set your "loc" value to your central value in stats.norm.pdf(). In regards to the width, modify np.linspace range and "scale" inside stats.norm.pdf(). You can also modify the x limits, though it may make it difficult to observe the change in width.

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d, 50, density = True)
plt.tick_params(labelsize = 24)
plt.xlim([2.95, 3.05])
x = np.linspace(2.975, 3.025, 1000)
ax.plot(x, stats.norm.pdf(x, loc = 3.0, scale = 0.005), linewidth = 8, alpha = 0.7)
plt.show()
```

When setting the y axis to LOG, make the same modifications as above (adjust the central value for plt.xlim(), np.linspace(), and stats.norm.pdf(), and the scale in np.linspace and stats.norm.pdf()). The shape of the plot confirms this remains a Gaussian distribution.

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d, 50, density = True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([2.95, 3.05])
x = np.linspace(2.975, 3.025, 1000)
ax.plot(x, stats.norm.pdf(x, loc = 3.0, scale = 0.005), linewidth = 8, alpha = 0.7)
plt.show()
```
